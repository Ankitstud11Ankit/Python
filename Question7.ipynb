```python
import pandas as pd
df=pd.DataFrame({
    "Employee":["Amit","Ravi","Neha","Amit","Priya","Ravi"],
    "Department":["IT","HR","IT","IT","Sales","HR"],
    "City":["Delhi","Mumbai","Delhi","Delhi","Pune","Mumbai"],
    "Salary":[60000,50000,70000,60000,55000,55000]
})
```

# Find unique department

```python
df["Department"].unique()
```

# Count employee by department

```python
df["Department"].value_counts()
```

# Find number of different cities

```python
df["City"].nunique()
```

# Find The average salary by department

```python
df.groupby("Department")["Salary"].mean()
```

Find The Highest Paid Employee

```python
df["Rank"]=df["Salary"].rank(ascending=False,method="dense")
df[df["Rank"]==1]
```

```python
df.loc[df["Salary"].idxmax()]
```
