```python
import pandas as pd
df=pd.read_csv(r"C:\Users\ankit\OneDrive\Desktop\employee_data.csv")
df["Salary"]=df["Salary"].replace({45000:40000})
df
```

```python
def grade(Performance):
    if Performance>=90:
        return "A"
    elif Performance>=80:
        return "B"
    elif Performance>=70:
        return "C"
    else:
        return "D"

df["Grade"]=df["Performance"].apply(grade)
df

```

```python
def salary(Salary):
    if Salary < 40000:
        return "low"
    elif Salary <= 70000 :
        return "Medium"
    else:
        return "High"

df["Salary Category"]=df["Salary"].apply(salary)
df
```

```python
df["Gender"]=df["Gender"].map({"M":"Male","F":"Female"}).str.strip()
df
```

```python
def expertise(Experience):
    if Experience<2:
        return "Fresher"
    elif Experience<=4:
        return "Mid-Level"
    else:
        return "senior"

df["Experience Level"]=df["Experience"].apply(expertise)
df
```

```python
df.to_csv(r"C:\Users\ankit\OneDrive\Desktop\Pandas\eployee.csv",index=False)
```
