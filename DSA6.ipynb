```python
import pandas as pd
df=pd.DataFrame({
    "Name":["A","B","C","D","E"],
    "Salary":[50000,70000,60000,90000,70000]

})
df
```

# Handiling ties with method

```python
# 1.method="average"--default
df["Rank"]=df["Salary"].rank(ascending=False,
                             method="average")
df
```

```python
# 2.method="min"
df["Minimum"]=df["Salary"].rank(ascending=False,
                                method="min")
df
```

```python
# 3.method="max"
df["Maximum"]=df["Salary"].rank(ascending=False,method="max")
df
```

```python
# 4.method="dense"
df["dense"]=df["Salary"].rank(ascending=False,method="dense")
df
```

```python
# 5.method="first"
df["First"]=df["Salary"].rank(ascending=False,method="dense")
df
```
