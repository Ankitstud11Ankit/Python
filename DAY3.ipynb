```python
import pandas as pd
data={
    "Name":["Amit","Rohit","Sneha","Priya","Karan"],
    "Age":[20,21,19,22,20],
    "Marks":[78,85,92,67,88]
}
df=pd.DataFrame(data)
df
```

# 1.apply(),

```python
df["Marks"]=df["Marks"].apply(lambda x:x+5) # Lammda means take x and return x+5 here
df
```

```python
def grade(Marks):
    if Marks>=90:
        return "A"
    elif Marks>=80:
        return "B"
    elif Marks>=70:
        return "C"
    else:
        return "D"
```

```python
df["Grade"]=df["Marks"].apply(grade)
df
```

# 2.map

```python
df["Gender"]=["M","M","F","F","M"]
```

```python
df["Gender"]=df["Gender"].map({"M":"Male","F":"Female"})
df
```

# Multiple Calculation

```python
df.groupby("Department")["Salary"].agg(['mean','max','min','count'])
```

Different calculation for different columns

```python
df.groupby("Department").agg({"Salary":["mean","max"],
                              "Performance":["mean","max"],
                              "Experience":"mean"})
```

Rename the results

```python
result=df.groupby("Department").agg(
    Average_Salary=("Salary","mean"),
    Maximum_Salary=("Salary","max"),
    Average_Performance=("Performance","mean"),
    Maximum_Performance=("Performance","max"),
    Average_Experience=("Experience","mean")

)
result
```

```python
employees=pd.DataFrame({
    "Employee_ID":[101,102,103,104,105],
    "Name":["Rahul","Priya","Aman","Neha","Rohit"],
    "Department ID":[1,2,1,3,2]
})

employees
```

```python
departments=pd.DataFrame({
    "Department ID":[1,2,3],
    "Departments":["IT","HR","Finance"]
})
departments
```

# Inner Join

```python
output=pd.merge(
    employees,departments,
    on="Department ID",
    how="inner"

)
output
```

# Left Join

```python
output=pd.merge(employees,departments,
                on="Department ID",
                how="left"
    
)
output
```

# Right Join

```python
output=pd.merge(employees,departments,
                on="Department ID",
                how="right"
    
)
output.sort_values("Employee_ID")
    
```

# Outer Join

```python
output=pd.merge(employees,departments,
                on="Department ID",
                how="outer")
output
```

# Concatination 

```python
df1=pd.DataFrame({"Name":["Rahul","Priya"],
                  "Salary":[45000,52000]})
df2=pd.DataFrame({"Name":["Aman","Neha"],
                  "Salary":[38000,75000]})
```

```python
result=pd.concat([df1,df2],ignore_index=True)
result
```

# Conactinating Horizontally

```python
df1=pd.DataFrame({"Name":["Rahul","Priya"],
                  "Salary":[45000,52000]})
df2=pd.DataFrame({"Experience":[2,3],
                 "Department":["IT","HR"]})
```

```python
result=pd.concat([df1,df2],axis=1)
result
```
