# Date & Time

```python
import pandas as pd
data=pd.DataFrame({"Name":["Rahul","Priya","Aman","Neha"],
                   "Joining Date":["2024-01-15","2023-06-20","2025-03-10","2022-11-05"]})
data.dtypes
```

# Convert to datetime

```python
data["Joining Date"]=pd.to_datetime(data["Joining Date"])
data.dtypes
```

# Extract the year

```python
data["Year"]=data["Joining Date"].dt.year
data
```

# Extract the month

```python
data["Month"]=data["Joining Date"].dt.month
data
```

# Get The Month Name

```python
data["Month Name"]=data["Joining Date"].dt.month_name()
data
```

# Extract Day

```python
data["Day"]=data["Joining Date"].dt.day
data
```

# **Find The Employees who joined after a date**

```python
data[data["Joining Date"]>"2024-01-01"]
```

# ** Find The Employee Who Joined in 2024**

```python
data[data["Joining Date"].dt.year==2024]
```

```python
data[data["Joining Date"].dt.year.isin([2024,2025])]
```

# Advance Merge And Join

** Merge When Column Names Are Different**

```python
# Suppose
employee=pd.DataFrame({
    "Employee ID":[101,102,103],
    "Name":["Rahul","Priya","Aman"],
    "Dept ID":[1,2,1]
})

departments=pd.DataFrame({
    "Department ID":[1,2],
    "Department":["IT","HR"]
})
```

```python
result=pd.merge(
    employee,
    departments,
    left_on="Dept ID",
    right_on="Department ID",
    how="left"
)
result
```

# Merge Uisng Multiple Column

```python
sales=pd.DataFrame({
    "Employee ID":[101,101,102],
    "Year":[2025,2026,2025],
    "Sales":[50000,65000,45000]
})

target=pd.DataFrame({
    "Employee ID":[101,101,102],
    "Year":[2025,2026,2025],
    "Target":[45000,60000,50000]
})
```

```python
pd.merge(
    sales,
    target,
    on=["Employee ID","Year"],
    how="left"
)
```

What if both tables having the same column name?

```python
sales=pd.DataFrame({
    "Employee ID":[101,101,102],
    "Year":[2025,2026,2025],
    "Sales":[50000,65000,45000]
})

target=pd.DataFrame({
    "Employee ID":[101,101,102],
    "Year":[2025,2026,2025],
    "Sales":[45000,60000,50000]
})
```

```python
result=pd.merge(
    sales,
    target,
    on=["Employee ID","Year"],
    suffixes=["_customers","_festival"]
    
)
result
```
