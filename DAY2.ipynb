```python
import pandas as pd
data={
    "Name":["  Amit","Rohit","Sneha","Priya","Amit","Ankit"],
    "City":["Delhi","delhi","Mumbai","Delhi","Delhi","mumbai"],
    "Marks":[78,85,92,67,78,87]

}
df=pd.DataFrame(data)
df
```

# To Find The Duplicate Value

```python
df.duplicated().sum()
```

# To Remove The Duplicate Value

```python
df.drop_duplicates(inplace=True)
df
```

# Inconsistent Values

```python
df["City"].value_counts()
```

# To Replace The value

```python
df["City"]=df["City"].replace("delhi","Delhi")
df
```

# To Replace the multiple value

```python
df["City"]=df["City"].replace({"delhi":"Delhi",
"mumbai":"Mumbai"})
df
```

# To Remove The extra spaces

```python
df["City"]=df["City"].str.strip()
df
```

```python
df["Name"]=df["Name"].str.strip()
df
```
