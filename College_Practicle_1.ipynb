```python
import pandas as pd
import numpy as np
data=pd.read_csv(r"C:\Users\ankit\OneDrive\Desktop\Company_Employee_Pandas_Practice.csv")
data.head(5)
```

```python
data.columns
```

```python
data.isnull().sum()
```

```python
# Handeling the missing Data 
data["Age"]=data["Age"].fillna(data["Age"].mean())
data["City"]=data["City"].fillna("Uknown")
data["Education"]=data["Education"].fillna("Unknown degree")
data["Salary"]=data["Salary"].fillna(data["Salary"].median())
data["Performance_Rating"]=data["Performance_Rating"].fillna("Uknown")
```

```python
# Handel The Inconsistent Values
data["Name"]=data["Name"].str.strip().str.lower()
```

```python
data.columns
```

```python
data["Gender"]=data["Gender"].str.strip().str.lower()
data["City"]=data["City"].str.strip().str.lower()
data["Department"]=data["Department"].str.strip().str.upper()
data["Job_Role"]=data["Job_Role"].str.strip().str.capitalize()
data["Education"]=data["Education"].str.strip().str.capitalize()
data["Status"]=data["Status"].str.strip().str.capitalize()

```
