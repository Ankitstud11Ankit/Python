# Advance Filtering

```python
import pandas as pd 
df=pd.read_csv(r"C:\Users\ankit\OneDrive\Desktop\employee_data.csv")
df
```

# isin --Filter Multiple values


```python
df[(df["Department"]=="IT")|(df["Department"]=="HR")]
```

Instead of that we can do this

```python
df[df["Department"].isin(["IT","HR"])]
```

# between --Filter with range

```python
df[df["Salary"].between(40000,70000)]
```

# Query 

```python
df[df["Salary"]>70000]
```

```python
df.query("Salary>70000")
```

# String operation

```python
# convert to lower case
df["Name"]=df["Name"].str.lower()
df["Name"]
```

```python
# Convert to Upper Case
df["Name"]=df["Name"].str.upper()
df["Name"]
```

```python
# Remove the space
df["Name"]=df["Name"].str.strip()
df["Name"]
```

```python
# To Find Name Conatining Any letter
df[df["Name"].str.contains('a',case=False)] # Case is for to ignore
```

```python
# Name Starting with
df[df["Name"].str.startswith("A")] # In this we can not use case argument
```

```python
# Name Endswith
df[df["Name"].str.endswith("T")]
```

```python
# To See the length
df["Length of Name"]=df["Name"].str.len()
df
```

```python
# To Repalce The Data
df["Department"]=df["Department"].str.replace('IT','Information Technology')
df
```
