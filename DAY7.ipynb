```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Ravi","Neha"],
    "Salary":[50000,60000,55000]
})
df
```

Index Change after filtering

```python
df[df["Salary"]>50000].reset_index(drop=True)
```

# To Set The Index to any column

```python
data=pd.DataFrame({
    "Employee ID":[101,102,103],
    "Name":["Amit","Ravi","Neha"],
    "Salary":[50000,60000,55000]
})
```

```python
df=data.set_index("Employee ID")
df
```

# nunique

```python
df=pd.DataFrame({
    "Name":["Amit","Ravi","Amit","Neha","Ravi",None],
    "City":["Delhi","Mumbai","Delhi","Pune","Mumbai",None],
    "Department":["IT","HR","IT","Sales","HR",None]
})
df
```

To see unique city

By default nunique does not count nun value

```python
df["City"].nunique()
```

if you want to count nan as a unique value

```python
df.nunique(dropna=False)
```

unique()=Actual Unique value

nunique()=Number of unique values

```python
df["City"].unique()
```
