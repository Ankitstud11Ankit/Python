# Date & Time

```python
import pandas as pd
df=pd.DataFrame({"Employee":["Rahul","Priya","Aman","Neha","Rohit"],
                 "Joining Date":["2022-05-10","2023-08-15","2024-01-20","2024-07-05","2025-03-12"]
                 })
df
```

```python
df["Joining Date"]=pd.to_datetime(df["Joining Date"])
df.dtypes
```

```python
df["Year"]=df["Joining Date"].dt.year
df

```

```python
df["Month Name"]=df["Joining Date"].dt.month_name()
df
```

```python
df[df["Joining Date"].dt.year.isin([2024])]
```

```python
df[df["Joining Date"]>"2024-12-31"]
```
